# Imports and variables

```python
import requests
import uuid

# URL used to connect to Fiddler
BASE_URL = ""

# Fiddler auth token
TOKEN = ""

# Request headers
HEADERS = {
    "Authorization": f"Bearer {TOKEN}",
    "Content-Type": "application/json"
}

# ML project and model IDs
ML_PROJECT_ID = ""
ML_MODEL_ID = ""

# Agentic project and application IDs
AGENTIC_PROJECT_ID = ""
AGENTIC_APPLICATION_ID = ""

# Metric retrieval endpoints
ML_MONITORING_METRICS_URL = f"{BASE_URL}/v3/queries"
ML_METRIC_CARD_METRICS_URL = f"{BASE_URL}/v3/analytics/metrics"
AGENTIC_MONITORING_METRICS_URL = f"{BASE_URL}/v3/queries"
AGENTIC_METRIC_CARD_METRICS_URL = f"{BASE_URL}/v3/queries"

START_TIME = "2026-02-03T00:00:00"
END_TIME = "2026-03-04T23:59:59"
```

# Payloads for each endpoints

```python

# Payload for retrieving monitoring metrics from ML models
ML_MONITORING_METRICS_PAYLOAD = {
    "project_id": ML_PROJECT_ID,
    "query_type":"MONITORING",
    "filters":{
        "bin_size":"Day",
        "time_zone":"UTC",
        "time_label":"30d",
        "time_range": {
            "start_time": START_TIME,
            "end_time": END_TIME
        }
    },
    "queries":[
        {
            "baseline_id":"",
            "columns":[],
            "metric":"traffic",
            "model_id": ML_MODEL_ID,
            "query_key": str(uuid.uuid4()),
            "segment":{},
            "viz_type":"line"
        }
  ]
}

# Payload for retrieving metric card metrics from ML models
ML_METRIC_CARD_METRICS_PAYLOAD = {
    "model_id": ML_MODEL_ID,
    "env_type":"PRODUCTION",
    "metrics":[
        {
            "id": "traffic"
        }
    ],
    "time_range":{
        "start_time": START_TIME,
        "end_time": END_TIME
    }
}

# Payload for retrieving monitoring metrics from agentic applications
AGENTIC_MONITORING_METRICS_PAYLOAD = {
    "project_id": AGENTIC_PROJECT_ID,
    "query_type": "GEN_AI_MONITORING",
    "time_filters": {
        "bin_size": "Day",
        "time_range": {
            "start_time": START_TIME,
            "end_time": END_TIME
        },
        "time_zone": "UTC"
    },"queries": [
        {
            "application_id": AGENTIC_APPLICATION_ID,
            "metric_name": "traffic",
            "metric_source": "raw_data",
            "query_key": str(uuid.uuid4()),
            "query_type": "GEN_AI_MONITORING",
            "query_scope": "SPAN",
            "aggregation": "count",
            "filters": {
                "agents": [],
                "span_groups": [],
                "span_types": [],
                "evaluator_rules": [],
                "token_counts": [],
                "attributes": [],
                "latency": []
            },
            "viz_type": "line"
        }
    ]
}

# Payload for retrieving metric card metrics from agentic applications
AGENTIC_METRIC_CARD_METRICS_PAYLOAD = {
    "project_id": AGENTIC_PROJECT_ID,
    "query_type": "GEN_AI_METRICS",
    "time_filters": {
        "time_range": {
            "start_time": START_TIME,
            "end_time": END_TIME
        },
        "time_zone": "UTC"
    },
    "queries":[
        {
            "aggregation":"count",
            "metric_name":"traffic",
            "metric_source":"raw_data",
            "query_scope":"SPAN",
            "query_key": str(uuid.uuid4()),
            "application_id": AGENTIC_APPLICATION_ID,
            "query_type":"GEN_AI_METRICS",
            "filters": {
                "agents":[],
                "span_groups":[],
                "span_types":[],
                "evaluator_rules":[],
                "token_counts":[],
                "attributes":[],
                "latency":[]
            },
            "viz_type":"line"
        }
    ]
}
```

# Retrieving monitoring metrics from an ML model

```python
response = requests.post(ML_MONITORING_METRICS_URL, headers=HEADERS, json=ML_MONITORING_METRICS_PAYLOAD)
response.raise_for_status()
data = response.json()
data["data"]["results"]
```

# Retrieving metric card metrics from an ML model

```python
response = requests.post(ML_METRIC_CARD_METRICS_URL, headers=HEADERS, json=ML_METRIC_CARD_METRICS_PAYLOAD)
response.raise_for_status()
data = response.json()
data["data"]
```

# Retrieving monitoring metrics from an agentic application

```python
response = requests.post(AGENTIC_MONITORING_METRICS_URL, headers=HEADERS, json=AGENTIC_MONITORING_METRICS_PAYLOAD)
response.raise_for_status()
data = response.json()
data["data"]["results"]
```

# Retrieving metric card metrics from an agentic application

```python
response = requests.post(AGENTIC_METRIC_CARD_METRICS_URL, headers=HEADERS, json=AGENTIC_METRIC_CARD_METRICS_PAYLOAD)
response.raise_for_status()
data = response.json()
data["data"]["results"]
```
